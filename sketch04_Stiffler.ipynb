```python
import pandas as pd

DATA_BASE = "https://raw.githubusercontent.com/pjbitterman/KSU_x9076/main/data"
oh_precip = pd.read_csv(f"{DATA_BASE}/part1/ohio_precip_2025.csv")
oh_precip.head()
```

```python
oh_precip["precip_mm_2025"].describe()
```

```python
top_5= oh_precip.sort_values("precip_mm_2025", ascending=False).head(5)[["name", "precip_mm_2025"]]
top_5.head()
```

```python
dist_from_mean= top_5.head(1)["precip_mm_2025"]-oh_precip["precip_mm_2025"].mean()
print(dist_from_mean)
```

The wettest station (1495.7mm) has 486.837561mm more rain than the mean precipitation amount (1008.862439mm). This seems like a significant difference especially considering that the maximum value is over 2 standared deviations away from the mean which makes it an unusual value.
